```python
import numpy as np
```

```python
state = np.array([0.2, -0.1, 0.03, -0.08])
weights = np.array([0, 0, 1, 0.5])
```

```python
# 기존 계산 방식
score1 = state[2] + 0.5 * state[3]
```

```python
# 내적을 사용하는 방식
score2 = np.dot(weights, state)
```

```python
# @ 연산자 사용
score3 = weights @ state
```

```python
print("Original:", score1)
print("Dot:", score2)
print("Matmul:", score3)
```

```python
print("State shape:", state.shape)
print("Weight shape:", weights.shape)
```
