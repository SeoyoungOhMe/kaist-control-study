```python
state = np.array([0.2, 0.5, 0.1, -0.3])
weights = np.array([1.0, 0.5, 2.0, 0.1])

print("State shape:", state.shape)
print("State dimension:", state.ndim)
print("Dot product:", np.dot(state, weights))
print("Elementwise:", state * weights)
```
